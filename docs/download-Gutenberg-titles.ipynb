```python
import os
from bs4 import BeautifulSoup as bsoup
import time

base_url = 'https://www.gutenberg.org/cache/epub/'
base_dir = '../data/Gutenberg/Nothern_Nut_Growers/'
for li in booklinks:
    # https://www.gutenberg.org/cache/epub/25703/pg25703.txt
    rel_ref = li.find('a').attrs['href']
    book_id = rel_ref.split('/')[-1]
    url = f"{base_url}{book_id}/pg{book_id}.txt"
    title = li.find('span', class_='title').text
    fname = os.path.join(base_dir, f"{title.replace(' ', '_')}.txt")
    print(url, fname)
    response = requests.get(url)
    if response.status_code == 200:
        with open(fname, 'wt') as fh:
            fh.write(response.text)
    time.sleep(10)
```
