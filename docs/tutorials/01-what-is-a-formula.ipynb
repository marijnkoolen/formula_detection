## A formulaic phrase, in plain terms

A **formulaic phrase** is a stretch of text that recurs, almost word-for-word, across many documents of the same kind &mdash; usually because the people writing those documents were following a convention, a legal requirement, or simply a habit of how things were "supposed" to be written.

You already know examples of this from your own life. A letter that begins "Dear Sir or Madam" and ends "Yours sincerely" is using formulaic phrases. A court verdict that says "the defendant is hereby found..." is using one. A wedding ceremony that says "do you take this person to be your..." is using one. None of these phrases are unique to one letter, one verdict, or one wedding &mdash; they're the *template* that countless individual instances were built from.

Historical administrative documents are often *full* of this kind of language, because the people writing them (clerks, secretaries, notaries) were trained to follow set conventions, often for legal or bureaucratic reasons. The example below is real: it's the opening of a 1768 resolution of the Dutch States General (a governing assembly), one of thousands of similar documents this toolkit has been tested on.

```python
example_1 = "Ontfangen een Missive van den Heere Gallieris, haar Hoog Mogende Minister by den Ryksdag, geschreeven te Regensburg den 28 der voorleede maand, houdende advertentie."
example_2 = "Ontfangen een Missive van den Heere van Kretschmar, haar Hoog Mogende extraordinaris Envoyé aan het Hof van sijne Majesteit den Koning van Sweeden, geschreeven te Stokholm den 28 der voorleede maand, houdende advertentie."

print(example_1)
print()
print(example_2)
```

Look closely at the two examples above (they're about two completely different people, on two different occasions). Notice how much of the wording is identical:

> Ontfangen een Missive van den Heere **[NAME]**, haar Hoog Mogende **[TITLE]**, geschreeven te **[PLACE]** den **[DATE]**, houdende advertentie.

("Received a letter from Lord [NAME], their High Mightinesses' [TITLE], written at [PLACE] on [DATE], containing notification.")

Everything outside the brackets is the **formula** &mdash; fixed wording that recurs across hundreds of these resolutions. Everything inside the brackets is a **variable slot** &mdash; the specific detail that changes every time the formula is used.

This is the central idea behind everything in this toolkit: **formulaic phrases are the stable scaffolding; the variable slots are where the actual historical content lives.** Once you can reliably tell the two apart, you can do things like:

* automatically pull out *who* sent a letter, *when*, and *from where*, across thousands of documents, without reading every single one by hand;
* notice when a clerk *deviates* from the usual formula &mdash; which is often where something historically interesting is happening;
* figure out where one document ends and the next begins, purely from the patterns in the formulas, even if you don't have that information already;
* track how formulaic language itself changes over decades or centuries &mdash; which is a historical question in its own right.

## Why not just read the formula off the page?

If you already know what the formula looks like, why do you need software for this at all? Two reasons, both very common in historical text:

**1. Spelling and wording drift.** The same clerk might write "Ontfangen" one week and "ontfangen" (lowercase) the next; a different clerk a generation later might write "Ontvangen". A scanned-and-transcribed document might contain recognition errors ("Ontfangsn"). All of these are, functionally, *the same formula* &mdash; but a simple search for the exact text "Ontfangen een Missive van" will miss most of them. We'll come back to this in tutorial 4.

**2. Scale.** You might be looking at one document and immediately recognise its formula. But if you have ten thousand documents spanning two centuries, reading each one to spot the pattern by hand simply isn't feasible. The whole point of this toolkit is to find these patterns automatically, across however much text you have, and to give you tools to inspect, confirm, and refine what it finds &mdash; rather than asking you to trust a black box.
